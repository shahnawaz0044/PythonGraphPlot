```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
```

```python
# sigle linear regression model
#house prediction dataset  
x = np.array([[1], [2], [3], [4], [5]]) #1 dimensional array
y = np.array([2, 4, 6, 8, 10])  #1 dimensional array/price

model = LinearRegression() #create model
model.fit(x, y) #train model

prediction = model.predict(np.array([[6]])) #predict price for 6

print("Predicted price for 6 is:", prediction[0])
```

```python
import matplotlib.pyplot as plt
x = np.array([1, 2, 3, 4, 5])
y = np.array([-2, 4, 6, 6, 10])
plt.plot(x, y, marker='o', linestyle='--', color='r', markerfacecolor='b', markersize=8)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Houses Data')
plt.savefig('plot0.png')  # Save the plot as a PNG file
plt.show()
```
